## 1. Pré-processamento de Imagens

### Introdução
O pré-processamento de imagens é uma etapa fundamental em projetos de visão computacional. Consiste em aplicar transformações nas imagens brutas antes de alimentá-las em modelos de aprendizado de máquina. O objetivo principal é melhorar a qualidade da imagem, remover ruídos, realçar características importantes e padronizar o formato (como redimensionamento e normalização), o que facilita e melhora o desempenho dos algoritmos subsequentes.

### Exemplos de bibliotecas/frameworks
- **OpenCV**: Uma das bibliotecas mais antigas e completas para visão computacional. Muito rápida e otimizada em C/C++ mas com excelentes interfaces em Python.
- **Pillow (PIL)**: Biblioteca padrão do Python para manipulação básica de imagens (abrir, salvar, recortar, redimensionar).
- **scikit-image**: Biblioteca baseada em SciPy, excelente para processamento de imagens mais voltado para pesquisa e manipulação de arrays NumPy.

### Exemplos de aplicações

```python
import cv2
import numpy as np

# Exemplo: Redimensionamento, conversão de cor e suavização.
# Criando uma imagem RGB aleatória para o exemplo (500x500)
imagem_bruta = np.random.randint(0, 255, (500, 500, 3), dtype=np.uint8)

# 1. Redimensionamento para o tamanho esperado por uma rede neural (ex: 224x224)
img_redimensionada = cv2.resize(imagem_bruta, (224, 224))

# 2. Conversão para escala de cinza 
img_cinza = cv2.cvtColor(img_redimensionada, cv2.COLOR_BGR2GRAY)

# 3. Suavização  usando filtro Gaussiano para remoção de ruídos
img_suavizada = cv2.GaussianBlur(img_cinza, (5, 5), 0)

print(f"Formato original: {imagem_bruta.shape}")
print(f"Formato processado final: {img_suavizada.shape}")
```

## 2. Segmentação de Imagens

### Introdução
A segmentação de imagens é o processo de particionar uma imagem em múltiplas regiões (conjuntos de pixels). Ao invés de classificar a imagem inteira, a segmentação atribui um rótulo a cada pixel da imagem, permitindo identificar a localização e os contornos exatos de diferentes objetos. É amplamente utilizada em imagens médicas (para identificar tumores em ressonâncias), carros autônomos (para identificar pedestres, faixas da via e outros carros) e edições de imagem que exigem recorte de fundo.

### Exemplos de bibliotecas/frameworks
- **Detectron2 (Facebook AI)**: Fornece modelos estado-da-arte como o Mask R-CNN para segmentação de instâncias.
- **TensorFlow/Keras e PyTorch**: Muito utilizados para implementar arquiteturas clássicas como a U-Net, excelente para segmentação semântica de imagens biomédicas.
- **OpenCV**: Oferece implementações de algoritmos clássicos de segmentação baseados em visão tradicional, como *Watershed* (divisão de bacias hidrográficas) e limiarização.

### Exemplos de aplicações

```python
# Exemplo simples de segmentação baseada em limiar (Thresholding) usando OpenCV
import cv2
import numpy as np

# Criando uma imagem sintética 
img = np.ones((200, 200), dtype=np.uint8) * 200
cv2.circle(img, (100, 100), 50, 50, -1) 

# Aplicando Binarização (threshold): 
# Pixels menores que 127 viram 0 
# Pixels maiores viram 255
_, mascara_segmentacao = cv2.threshold(img, 127, 255, cv2.THRESH_BINARY_INV)
```

## 3. Detecção/Classificação de Imagens

### Introdução
A **Classificação** de imagens tenta responder à pergunta "o que é isso?" analisando uma imagem como um todo e atribuindo-lhe um rótulo (ex: identificando se uma foto é de um gato ou um cachorro). A **Detecção** de objetos dá um passo além, tentando responder "o que há na imagem e onde estão os objetos?", fornecendo não só a classe, mas também desenhando caixas delimitadoras (bounding boxes) na localização exata dos objetos. Essas tarefas são a essência do reconhecimento facial, leitura de placas e indexação de bibliotecas de fotos.

### Exemplos de bibliotecas/frameworks
- **YOLO (You Only Look Once)**: Família de modelos extremamente rápidos para detecção de objetos em tempo real (ex: YOLOv8 via pacote `ultralytics`).
- **TensorFlow / Keras**: Facilita o carregamento de redes de classificação robustas e prontas para uso, como ResNet50, Inception e MobileNet.
- **Torchvision (PyTorch)**: Oferece módulos para carregar modelos pré-treinados de classificação e detecção de objetos como Faster R-CNN e RetinaNet.

### Exemplos de aplicações

### Exemplos teóricos de aplicações na vida real

- **Reconhecimento Facial (Autenticação)**: Utilizado amplamente em segurança de smartphones (como o FaceID da Apple) e em sistemas bancários para evitar fraudes, onde o modelo de classificação verifica se as características do rosto detectado na câmera correspondem às do usuário cadastrado.
- **Veículos Autônomos (Detecção em tempo real)**: Carros da Tesla e Waymo utilizam modelos complexos (como variantes do YOLO) para detectar, localizar com bounding boxes e classificar múltiplos objetos simultaneamente nas ruas: pedestres, placas de trânsito, semáforos e outros carros.
- **Diagnóstico Médico por Imagem**: Hospitais utilizam sistemas de classificação para analisar raios-x e ressonâncias magnéticas. A IA pode classificar uma radiografia de pulmão indicando a probabilidade de haver pneumonia, auxiliando os médicos em diagnósticos mais rápidos.
- **Agricultura de Precisão**: Drones sobrevoam plantações tirando fotos aéreas, e modelos de detecção identificam e localizam ervas daninhas no meio da lavoura, permitindo a aplicação de defensivos agrícolas apenas onde é estritamente necessário.
